Para introduzir as teorias que serão analisadas nesta tarefa:

* Distribuição Normal:Uma distribuição contínua simétrica em formato de sino, definida pela média ($\mu$) e pelo desvio padrão (σ) .
Utilização: Base para métodos paramétricos e modelagem de fenômenos naturais (alturas, erros de medida), servindo de referência para as demais distribuições.
* Distribuição t de student: É uma variação da distribuição Normal usada para amostras pequenas ou quando não conhecemos o desvio-padrão da população (precisando estimá-lo a partir da amostra).O formato é simétrico como a Normal, mas com caudas mais largas (achatadas). Isso reflete a incerteza adicional causada pelo tamanho reduzido da amostra.Conforme o tamanho da amostra aumenta, a distribuição $t$ se aproxima e se transforma exatamente na distribuição Normal Standard isso é chamada de Graus de Liberdade ($df$) representada pela equação  (\nu= n - 1\))
* Distribuição Qui-Quadrado ($\chi^2$): Surge da soma de variáveis normais padronizadas elevadas ao quadrado. Por envolver apenas valores ao quadrado, ela é composta apenas por valores não negativos ($x \ge 0$).Possui formato assimétrico (inclinado à direita). A curva muda de formato dependendo dos seus graus de liberdade e é utilizada para analisar frequências de variáveis categóricas ou testar a variância de uma população.
* Distribuição F de Fisher: É gerada pela razão entre duas distribuições Qui-Quadrado independentes, cada uma dividida pelos seus respectivos graus de liberdade. Como é uma razão de valores ao quadrado, também assume apenas valores não negativos ($F \ge 0$). Possui formato assimétrico à direita, dependendo de dois conjuntos de graus de liberdade ($df_1$ no numerador e $df_2$ no denominador).Pode ser utilizada para comparar duas variâncias ou razões de variação e usualmente é aplicada em ANOVA (Análise de Variância), para testar se há diferença significativa entre as médias de três ou mais grupos simultaneamente, ou em regressão linear,para avaliar se o modelo de regressão construído é estatisticamente significativo como um todo.

As fórmulas de cada distribuição estão descritas abaixo:
------------------------------
## 1. Distribuição Normal
A equação descreve a clássica curva em forma de sino, centralizada na média populacional.
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x - \mu}{\sigma}\right)^2}$$

*
* Domínio: $-\infty < x < \infty$
* $\mu$ (Mú): Média da distribuição (localização do centro).
* $\sigma$ (Sigma): Desvio padrão (indica a dispersão ou largura do sino).
* $e$ e $\pi$: Constantes matemáticas.
*

------------------------------
## 2. Distribuição t de Student
Esta fórmula utiliza a Função Gama ($\Gamma$) para ajustar as caudas mais grossas em amostras pequenas.
$$f(t) = \frac{\Gamma\left(\frac{\nu + 1}{2}\right)}{\sqrt{\nu\pi} \,\, \Gamma\left(\frac{\nu}{2}\right)} \left(1 + \frac{t^2}{\nu}\right)^{-\frac{\nu + 1}{2}}$$

*
* Domínio: $-\infty < t < \infty$
* $\nu$ (Nu): Graus de liberdade ($\nu = n - 1$).
* $\Gamma(z)$: Função Gama, que funciona como uma extensão do fatorial para números reais (onde $\Gamma(n) = (n-1)!$ para inteiros).
*

------------------------------
## 3. Distribuição Qui-Quadrado ($\chi^2$)
Como ela é baseada na soma de variáveis ao quadrado, assume apenas valores maiores ou iguais a zero.
$$f(x) = \frac{1}{2^{\frac{\nu}{2}} \Gamma\left(\frac{\nu}{2}\right)} x^{\frac{\nu}{2} - 1} e^{-\frac{x}{2}}$$

*
* Domínio: $x \ge 0$
* $\nu$: Graus de liberdade da distribuição.
* $e$: Base do logaritmo natural.
*

------------------------------
## 4. Distribuição F de Fisher
A fórmula modela a razão entre duas variâncias amostrais e, por isso, depende de dois graus de liberdade diferentes.
$$f(f) = \frac{\Gamma\left(\frac{\nu_1 + \nu_2}{2}\right)}{\Gamma\left(\frac{\nu_1}{2}\right) \Gamma\left(\frac{\nu_2}{2}\right)} \left(\frac{\nu_1}{\nu_2}\right)^{\frac{\nu_1}{2}} f^{\frac{\nu_1}{2} - 1} \left(1 + \frac{\nu_1}{\nu_2}f\right)^{-\frac{\nu_1 + \nu_2}{2}}$$

*
* Domínio: $f \ge 0$
* $\nu_1$: Graus de liberdade do numerador (associados à primeira variância).
* $\nu_2$: Graus de liberdade do denominador (associados à segunda variância).
