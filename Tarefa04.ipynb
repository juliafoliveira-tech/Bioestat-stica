```python
import pandas as pd

df = pd.read_excel('Tarefa04(1).xlsx')
```

```python
# Recarrega o arquivo Excel, usando a segunda linha (índice 1) como cabeçalho.
df_corrected = pd.read_excel('Tarefa04(1).xlsx', header=1)

# Exibe as primeiras linhas e informações do DataFrame corrigido para verificação.
display(df_corrected.head())
display(df_corrected.info())
```

```python
# Seleciona apenas as colunas numéricas do DataFrame
numerical_cols = df_corrected.select_dtypes(include=['number'])

if not numerical_cols.empty:
    # Calcula o primeiro quartil (25º percentil) para cada coluna numérica
    first_quartile = numerical_cols.quantile(0.25)
    print("Primeiro quartil para cada coluna numérica do arquivo 'Tarefa04(1).xlsx':")
    display(first_quartile)
else:
    print("Não foram encontradas colunas numéricas no DataFrame para calcular o quartil.")
```

### Interpretação do Primeiro Quartil

Com base nos valores calculados do primeiro quartil:

*   **Casos:** O primeiro quartil para 'Casos' é **2.0**. Isso significa que 25% dos locais registrados tiveram **2 casos ou menos**. Em outras palavras, a maioria (75%) dos locais teve mais de 2 casos.

*   **Surtos:** O primeiro quartil para 'Surtos' é **1.0**. Isso indica que 25% dos locais registrados tiveram **1 surto ou menos**. A maioria (75%) dos locais teve mais de 1 surto.

Em resumo, o primeiro quartil nos ajuda a entender o limite inferior para a maioria dos dados, excluindo os 25% de valores mais baixos.

```python
# Seleciona as colunas numéricas do DataFrame corrigido
numerical_cols_corrected = df_corrected.select_dtypes(include=['number'])

if not numerical_cols_corrected.empty:
    # Calcula o segundo quartil (mediana, 50º percentil) para cada coluna numérica
    second_quartile = numerical_cols_corrected.quantile(0.50)
    print("Segundo quartil (mediana) para cada coluna numérica do arquivo 'Tarefa04(1).xlsx':")
    display(second_quartile)
else:
    print("Não foram encontradas colunas numéricas no DataFrame para calcular o segundo quartil.")
```

### Interpretação do Segundo Quartil (Mediana)

Com base nos valores calculados do segundo quartil (mediana):

*   **Casos:** A mediana para 'Casos' é **4.0**. Isso significa que 50% dos locais registrados tiveram **4 casos ou menos**, e os outros 50% tiveram mais de 4 casos. É o valor central da distribuição de casos.

*   **Surtos:** A mediana para 'Surtos' é **1.0**. Isso indica que 50% dos locais registrados tiveram **1 surto ou menos**, e os outros 50% tiveram mais de 1 surto. É o valor central da distribuição de surtos.

Em resumo, a mediana divide os dados em duas metades iguais, indicando o ponto onde metade dos valores são menores e metade são maiores.

Agora, vamos calcular o terceiro quartil (75º percentil) para as colunas numéricas 'Casos' e 'Surtos' no DataFrame `df_corrected`.

```python
# Calcula o terceiro quartil (75º percentil) para as colunas numéricas
third_quartile = df_corrected.select_dtypes(include=['number']).quantile(0.75)

print("Terceiro quartil para cada coluna numérica do arquivo 'Tarefa04(1).xlsx':")
display(third_quartile)
```

### Interpretação do Terceiro Quartil

Com base nos valores calculados do terceiro quartil:

*   **Casos:** O terceiro quartil para 'Casos' é **30.0**. Isso significa que 75% dos locais registrados tiveram **30 casos ou menos**. Apenas 25% dos locais tiveram mais de 30 casos.

*   **Surtos:** O terceiro quartil para 'Surtos' é **6.0**. Isso indica que 75% dos locais registrados tiveram **6 surtos ou menos**. Apenas 25% dos locais tiveram mais de 6 surtos.

Em resumo, o terceiro quartil nos ajuda a entender o limite superior para a maioria dos dados, excluindo os 25% de valores mais altos.

Vamos criar um gráfico de boxplot para visualizar a distribuição dos dados das colunas 'Casos' e 'Surtos', incluindo seus quartis.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

# Boxplot para a coluna 'Casos'
plt.subplot(1, 2, 1) # 1 linha, 2 colunas, 1º gráfico
sns.boxplot(y=df_corrected['Casos'], palette='Blues')
plt.title('Boxplot de Casos')
plt.ylabel('Número de Casos')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Boxplot para a coluna 'Surtos'
plt.subplot(1, 2, 2) # 1 linha, 2 colunas, 2º gráfico
sns.boxplot(y=df_corrected['Surtos'], palette='Greens')
plt.title('Boxplot de Surtos')
plt.ylabel('Número de Surtos')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Interpretação do Boxplot

No boxplot, você pode observar:

*   **Linha inferior da caixa:** Representa o primeiro quartil (Q1).
    *   Para **Casos**, Q1 é **2.0**, significando que 25% dos locais tiveram 2 casos ou menos.
    *   Para **Surtos**, Q1 é **1.0**, significando que 25% dos locais tiveram 1 surto ou menos.

*   **Linha do meio da caixa:** Representa o segundo quartil (mediana ou Q2).
    *   Para **Casos**, a mediana é **4.0**, indicando que metade dos locais teve 4 casos ou menos.
    *   Para **Surtos**, a mediana é **1.0**, indicando que metade dos locais teve 1 surto ou menos.

*   **Linha superior da caixa:** Representa o terceiro quartil (Q3).
    *   Para **Casos**, Q3 é **30.0**, significando que 75% dos locais tiveram 30 casos ou menos.
    *   Para **Surtos**, Q3 é **6.0**, significando que 75% dos locais tiveram 6 surtos ou menos.

*   **"Bigodes" (whiskers):** Estendem-se até o menor e o maior valor dentro de 1.5 vezes o intervalo interquartil (IQR), que é a diferença entre Q3 e Q1.
*   **Pontos fora dos bigodes:** Representam valores atípicos (outliers), que são observações que se desviam significativamente da maioria dos dados.

Este gráfico oferece uma visão rápida da dispersão, simetria e presença de outliers nas distribuições de 'Casos' e 'Surtos'. Para 'Casos', a caixa do boxplot é mais alongada e com o 'bigode' superior bem extenso, indicando uma maior variabilidade e a presença de vários outliers no lado superior (valores altos de casos). Para 'Surtos', a caixa é mais compacta e o 'bigode' inferior é bem curto, sugerindo que a maioria dos valores está concentrada em um número menor de surtos, embora também haja outliers no lado superior.

### Como determinamos se há observações atípicas em uma base de dados?
### A partir da observação dos pontos fora é possível analisar os valores atípicos, ou seja, desviam dos outros dados e devem ser observados com mais cautela para aferir o resultado.
