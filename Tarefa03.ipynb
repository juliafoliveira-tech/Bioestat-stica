```python
frequency_table = dados.groupby(['Fenotipo_Sanguineo (Nominal)']).size().reset_index(name='Frequência Absoluta')
frequency_table['Frequência Relativa (%)'] = (frequency_table['Frequência Absoluta'] / len(dados)) * 100

fig = px.bar(
    frequency_table,
    x='Fenotipo_Sanguineo (Nominal)',
    y='Frequência Absoluta',
    title='Frequência Absoluta de Fenótipo Sanguíneo',
    labels={'Fenotipo_Sanguineo (Nominal)': 'Fenótipo Sanguíneo', 'Frequência Absoluta': 'Número de Indivíduos'}
)
fig.update_layout(xaxis_categoryorder='total descending')
fig.show()
```

###Esse gráfico de barras mostra a frequência de cada tipo de fenótipo sanguíneo na amostra. É possível observar a proporção de indivíduos para cada grupo sanguíneo.

```python
fig = px.histogram(
    dados,
    x='Acertos_Exame_Botanica (Discreta)',
    nbins=len(dados['Acertos_Exame_Botanica (Discreta)'].unique()),
    title='Distribuição de Acertos no Exame de Botânica',
    labels={'Acertos_Exame_Botanica (Discreta)': 'Número de Acertos', 'count': 'Frequência'}
)
fig.update_layout(xaxis=dict(tickmode='linear', dtick=1))
fig.show()
```

### O histograma exibe a distribuição da quantidade de acertos no exame de Botânica. Podemos ver quais pontuações são mais comuns entre os estudantes.

```python
fig = px.strip(
    dados,
    x='Fenotipo_Sanguineo (Nominal)',
    y='Acertos_Exame_Botanica (Discreta)',
    color='Fenotipo_Sanguineo (Nominal)', # Colore os pontos pelo tipo sanguíneo
    title='Acertos no Exame de Botânica por Fenótipo Sanguíneo',
    labels={
        'Fenotipo_Sanguineo (Nominal)': 'Fenótipo Sanguíneo',
        'Acertos_Exame_Botanica (Discreta)': 'Número de Acertos'
    },
    hover_data=['Nome_Ficticio', 'ID_Aluno'] # Mostra nome e ID ao passar o mouse
)
fig.show()
```

###Este gráfico de dispersão (strip plot) relaciona o fenótipo sanguíneo com a quantidade de acertos no exame de Botânica. Ele nos permite visualizar como os acertos se distribuem para cada grupo sanguíneo e se há alguma tendência ou diferença notável entre eles.
