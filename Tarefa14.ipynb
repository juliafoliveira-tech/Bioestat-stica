Exercício 1 — Massa corporal de mamíferos
A massa corporal de uma espécie de pequeno mamífero apresenta distribuição normal:
* μ=420 g
* σ=55 g
Determine:
1. A probabilidade de um indivíduo pesar menos de 350 g.
2. A probabilidade de pesar mais de 500 g.
3. A probabilidade de pesar entre 400 e 480 g.
4. Represente graficamente as três probabilidades

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da distribuição normal
mu = 420
sigma = 55

# 1. Probabilidade de pesar menos de 350 g
p1 = norm.cdf(350, loc=mu, scale=sigma)
print(f"1. Probabilidade de pesar menos de 350g: {p1:.4f} ({p1*100:.2f}%)")

# 2. Probabilidade de pesar mais de 500 g
p2 = 1 - norm.cdf(500, loc=mu, scale=sigma)
print(f"2. Probabilidade de pesar mais de 500g: {p2:.4f} ({p2*100:.2f}%)")

# 3. Probabilidade de pesar entre 400 e 480 g
p3 = norm.cdf(480, loc=mu, scale=sigma) - norm.cdf(400, loc=mu, scale=sigma)
print(f"3. Probabilidade de pesar entre 400 e 480g: {p3:.4f} ({p3*100:.2f}%)")
```

```python
# 4. Representação gráfica das três probabilidades
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, loc=mu, scale=sigma)

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# Gráfico 1: P(X < 350)
axes[0].plot(x, y, 'r-', lw=2)
px1 = np.linspace(mu - 4*sigma, 350, 500)
axes[0].fill_between(px1, norm.pdf(px1, loc=mu, scale=sigma), color='red', alpha=0.3,
                     label=f'P(X < 350) = {p1:.4f}')
axes[0].set_title("Pessando menos de 350g")
axes[0].set_xlabel("Massa (g)")
axes[0].set_ylabel("Densidade de Probabilidade")
axes[0].axvline(350, color='red', linestyle='--')
axes[0].legend(loc='upper right')
axes[0].grid(True, linestyle=':', alpha=0.6)

# Gráfico 2: P(X > 500)
axes[1].plot(x, y, 'b-', lw=2)
px2 = np.linspace(500, mu + 4*sigma, 500)
axes[1].fill_between(px2, norm.pdf(px2, loc=mu, scale=sigma), color='blue', alpha=0.3,
                     label=f'P(X > 500) = {p2:.4f}')
axes[1].set_title("Pessando mais de 500g")
axes[1].set_xlabel("Massa (g)")
axes[1].axvline(500, color='blue', linestyle='--')
axes[1].legend(loc='upper right')
axes[1].grid(True, linestyle=':', alpha=0.6)

# Gráfico 3: P(400 < X < 480)
axes[2].plot(x, y, 'g-', lw=2)
px3 = np.linspace(400, 480, 500)
axes[2].fill_between(px3, norm.pdf(px3, loc=mu, scale=sigma), color='green', alpha=0.3,
                     label=f'P(400 < X < 480) = {p3:.4f}')
axes[2].set_title("Pessando entre 400g e 480g")
axes[2].set_xlabel("Massa (g)")
axes[2].axvline(400, color='green', linestyle='--')
axes[2].axvline(480, color='green', linestyle='--')
axes[2].legend(loc='upper right')
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.suptitle("Distribuição Normal da Massa Corporal de Mamíferos (μ=420g, σ=55g)", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

Exercício 2 — Percentis de tamanho corporal
O comprimento corporal desses mamíferos segue uma distribuição normal com:
* μ=150 cm
* σ=10,5 cm
Calcule:
1. O percentil 80.
2. O valor correspondente a 10% dos indivíduos apresentarem o comprimento superior a esse
valor.
3. Faça um único gráfico indicando os três percentis.
Interprete biologicamente cada resultado. Por exemplo: qual comprimento separa os 10% maiores indivíduos dos demais?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da distribuição normal para comprimento corporal
mu_comp = 150
sigma_comp = 10.5

# 1. Percentil 80 (p = 0.80)
p_80 = norm.ppf(0.80, loc=mu_comp, scale=sigma_comp)

# 2. Valor correspondente a 10% com comprimento superior (p = 0.90)
p_90 = norm.ppf(0.90, loc=mu_comp, scale=sigma_comp)

print(f"1. Percentil 80: {p_80:.2f} cm")
print(f"2. Comprimento que separa os 10% maiores (Percentil 90): {p_90:.2f} cm")

# 3. Gráfico único mostrando os percentis
x_comp = np.linspace(mu_comp - 4*sigma_comp, mu_comp + 4*sigma_comp, 1000)
y_comp = norm.pdf(x_comp, loc=mu_comp, scale=sigma_comp)

plt.figure(figsize=(10, 6))
plt.plot(x_comp, y_comp, 'k-', lw=2, label='Distribuição do Comprimento')

# Linha do Percentil 80
plt.axvline(p_80, color='orange', linestyle='--', lw=2, label=f'Percentil 80 ({p_80:.2f} cm)')
plt.fill_between(x_comp, y_comp, where=(x_comp <= p_80), color='orange', alpha=0.15)

# Linha do Percentil 90
plt.axvline(p_90, color='red', linestyle=':', lw=2, label=f'Percentil 90 ({p_90:.2f} cm)')
plt.fill_between(x_comp, y_comp, where=(x_comp > p_90), color='red', alpha=0.15)

# Estilização do gráfico
plt.title("Distribuição do Comprimento Corporal com Percentis de Destaque", fontsize=14)
plt.xlabel("Comprimento Corporal (cm)", fontsize=12)
plt.ylabel("Densidade de Probabilidade", fontsize=12)
plt.legend(loc='upper left')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

Interpretação biológica: O comprimento de 163,46 cm separa os 10% maiores indivíduos (morfotipos maiores ou indivíduos mais velhos/maduros) do restante da população.

Exercício 3 — Integração
Uma população de aves apresenta massa corporal normalmente distribuída com:
* μ=72 g
* σ=8 g
Em Python:
1. Calcule P(X≤60).
2. Calcule P(X≥85).
3. Calcule P(65≤X≤80).
4. Calcule os percentis 90, 95 e 99.
5. Produza um gráfico da distribuição normal mostrando as três regiões de probabilidade e os três percentis.
6. Escreva uma interpretação ecológica curta dos resultados.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros
mu_aves = 72
sigma_aves = 8

# 1. Calculando P(X <= 60)
p_menor_60 = norm.cdf(60, loc=mu_aves, scale=sigma_aves)

# 2. Calculando P(X >= 85)
p_maior_85 = 1 - norm.cdf(85, loc=mu_aves, scale=sigma_aves)

# 3. Calculando P(65 <= X <= 80)
p_65_80 = norm.cdf(80, loc=mu_aves, scale=sigma_aves) - norm.cdf(65, loc=mu_aves, scale=sigma_aves)

# 4. Calculando os percentis 90, 95 e 99
pct_90 = norm.ppf(0.90, loc=mu_aves, scale=sigma_aves)
pct_95 = norm.ppf(0.95, loc=mu_aves, scale=sigma_aves)
pct_99 = norm.ppf(0.99, loc=mu_aves, scale=sigma_aves)

print("=== RESULTADOS NUMÉRICOS ===")
print(f"1. P(X <= 60): {p_menor_60:.4f} ({p_menor_60*100:.2f}%)")
print(f"2. P(X >= 85): {p_maior_85:.4f} ({p_maior_85*100:.2f}%)")
print(f"3. P(65 <= X <= 80): {p_65_80:.4f} ({p_65_80*100:.2f}%)")
print(f"4. Percentil 90: {pct_90:.2f} g")
print(f"   Percentil 95: {pct_95:.2f} g")
print(f"   Percentil 99: {pct_99:.2f} g\n")

# 5. Gráfico único mostrando as três regiões e percentis
x_aves = np.linspace(mu_aves - 4*sigma_aves, mu_aves + 4*sigma_aves, 1000)
y_aves = norm.pdf(x_aves, loc=mu_aves, scale=sigma_aves)

plt.figure(figsize=(14, 8))
plt.plot(x_aves, y_aves, 'k-', lw=2.5, label='Distribuição de Massa (Aves)')

# Sombreando as três regiões de probabilidade
plt.fill_between(x_aves, y_aves, where=(x_aves <= 60), color='lightblue', alpha=0.5,
                 label=f'P(X <= 60) = {p_menor_60*100:.2f}%')
plt.fill_between(x_aves, y_aves, where=(x_aves >= 85), color='lightcoral', alpha=0.5,
                 label=f'P(X >= 85) = {p_maior_85*100:.2f}%')
plt.fill_between(x_aves, y_aves, where=((x_aves >= 65) & (x_aves <= 80)), color='lightgreen', alpha=0.4,
                 label=f'P(65 <= X <= 80) = {p_65_80*100:.2f}%')

# Adicionando linhas verticais para os percentis
plt.axvline(pct_90, color='orange', linestyle='--', lw=2, label=f'Percentil 90 ({pct_90:.1f} g)')
plt.axvline(pct_95, color='darkorange', linestyle='-.', lw=2, label=f'Percentil 95 ({pct_95:.1f} g)')
plt.axvline(pct_99, color='red', linestyle=':', lw=2, label=f'Percentil 99 ({pct_99:.1f} g)')

# Customizações estéticas
plt.title("Distribuição da Massa Corporal da População de Aves", fontsize=16, pad=15)
plt.xlabel("Massa Corporal (g)", fontsize=12)
plt.ylabel("Densidade de Probabilidade", fontsize=12)
plt.xlim(mu_aves - 3.5*sigma_aves, mu_aves + 3.5*sigma_aves)
plt.legend(loc='upper left', fontsize=10, frameon=True, facecolor='white')
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

A grande maioria da população de aves (cerca de $65\%$) concentra-se em uma faixa intermediária de massa corporal entre $65\text{ g}$ e $80\text{ g}$, o que reflete o padrão fenotípico ótimo adaptado às condições ambientais e à disponibilidade de recursos da espécie. Indivíduos nas caudas da distribuição são raros: apenas $6.68\%$ apresentam massas inferiores a $60\text{ g}$ (o que pode indicar filhotes tardios, menor aptidão física ou restrição alimentar), enquanto apenas $5.21\%$ ultrapassam $85\text{ g}$, representando indivíduos excepcionalmente robustos ou em condições de sobrealimentação. Os percentis elevados mostram que apenas $1\%$ da população atinge massas superiores a $90.61\text{ g}$.
